# Diagnostic Variance Inflation Factor (VIF) Analysis (Tables 1, 3, 6)
This notebook calculates VIF values to diagnose collinearity in raw predictors, motivating the use of PCA.


```python
import sys
sys.path.append('..')

import pandas as pd
from sklearn.preprocessing import StandardScaler
from statsmodels.stats.outliers_influence import variance_inflation_factor
from src.data_loader import generate_simulation_data, load_concrete_dataset, load_energy_dataset
```

```python
def compute_vif(X, names):
    scaler = StandardScaler()
    X_sc = scaler.fit_transform(X)
    vifs = [variance_inflation_factor(X_sc, i) for i in range(X_sc.shape[1])]
    return pd.DataFrame({'Feature': names, 'VIF': vifs})
```

### 1. Simulation VIF (Table 1)


```python
X_sim, _, _ = generate_simulation_data(n=500, seed=42)
compute_vif(X_sim, ['X1', 'X2', 'X3', 'X4'])
```

### 2. Concrete VIF (Table 3)


```python
X_conc, _, c_names = load_concrete_dataset()
compute_vif(X_conc, c_names)
```

### 3. Energy VIF (Table 6)


```python
X_en, _ = load_energy_dataset('../data/enb2012.xlsx')
compute_vif(X_en, [f'X{i+1}' for i in range(X_en.shape[1])])
```
